```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
g = lambda x: (x+1)*(x+2)
def spots(n):
    sum = 0
    while n >= 0:
        sum += g(n)
        n -= 1
    return sum
```

```python
n_range = range(0,6)
for n in n_range:
    plt.axhline(spots(n), color="black")
    plt.yticks([spots(n) for n in n_range])
plt.xticks([])
plt.ylim([-10, 120])
plt.axhline(0, color="black", linestyle = "dashed", alpha = 0.5)
plt.ylabel("Number of neutrons (or protons)")
plt.savefig(r"C:\Users\Bonsaibäumchen\Desktop\bachelorarbeit\overleaf bilder\magicnoorbit.png", dpi = 300)
```
