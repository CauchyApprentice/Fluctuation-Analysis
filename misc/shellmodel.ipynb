```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
V0 = 1.6
R = 1.0
a = 0.2
k = 1.5

saxon = lambda x: -V0/(1+np.exp((x-R)/a))
harm = lambda x: 1/2 * k * np.square(x) - V0
```

```python
x_range = np.arange(-0.5,R+2,0.01)
saxon_range = [saxon(x) for x in x_range]
harm_range = [harm(x) for x in x_range]
```

```python
plt.plot(x_range,saxon_range, label = "Woods-Saxon")
plt.plot(x_range, harm_range, label = "Harm. Osc.")
plt.legend()
plt.xlabel("Radius")
plt.ylabel("Potential")
plt.xticks([0, R], [0, "R"])
plt.yticks([-V0, 0], ["-V0", 0])
plt.xlim(-0.2,2.5)
plt.ylim(-V0,0.5)

plt.savefig(r"C:\RAINIER\sample_folder\saxonharm.png", dpi = 300)
```
